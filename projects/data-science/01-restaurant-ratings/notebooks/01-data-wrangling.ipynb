### Cruzando as informações das tabelas

```python
nulos_restaurant_ratings = (
    ratings
    .merge(consumers, on="Consumer_ID", how="left")
    .merge(restaurant_cuisines, on="Restaurant_ID", how="left")
    .merge(restaurants, on="Restaurant_ID", how="left")
    .merge(consumer_preferences, on="Consumer_ID", how="left")
    .reset_index()
    [['Smoker', 'Transportation_Method', 'Marital_Status', 'Children', 'Occupation', 'Budget', 'Zip_Code', 'Alcohol_Service', 'Parking']]
    .head(5)
)
display(nulos_restaurant_ratings)
```

### Iniciando o tratamento dos dados

```python
consumer_preferences = consumer_preferences.drop_duplicates()

restaurants['Zip_Code'] = restaurants['Zip_Code'].fillna(restaurants['Zip_Code'].median())
restaurants['Alcohol_Service'] = restaurants['Alcohol_Service'].fillna('Não Informado')
restaurants['Parking'] = restaurants['Parking'].fillna('Não Informado')

consumers['Smoker'] = consumers['Smoker'].fillna('Não Informado')
consumers['Transportation_Method'] = consumers['Transportation_Method'].fillna('Não Informado')
consumers['Marital_Status'] = consumers['Marital_Status'].fillna('Não Informado')
consumers['Children'] = consumers['Children'].fillna('Não Informado')
consumers['Occupation'] = consumers['Occupation'].fillna('Não Informado')
consumers['Budget'] = consumers['Budget'].fillna('Não Informado')
```

### Exportando os arquivos tratados

```python
consumer_preferences.to_csv('../data/processed/consumer_preferences.csv', index=False)
consumers.to_csv('../data/processed/consumers.csv', index=False)
ratings.to_csv('../data/processed/ratings.csv', index=False)
restaurant_cuisines.to_csv('../data/processed/restaurant_cuisines.csv', index=False)
restaurants.to_csv('../data/processed/restaurants.csv', index=False)
```

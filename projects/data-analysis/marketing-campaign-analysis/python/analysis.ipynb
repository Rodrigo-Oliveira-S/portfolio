```python
from database import get_engine
from queries import (
    total_vendas, vendas_mes, crescimento, 
    consumo, campanha, pais, tempo_relacionamento,
    faixa_etaria, reclamacao
)
from sqlalchemy import text
import pandas as pd

engine = get_engine()

mes='2012-12'
```

```python
df1 = pd.read_csv('../raw/marketing_data.csv')
df2 = pd.read_csv('../raw/marketing_data_dictionary.csv')

df1.to_sql('marketing_data', engine, if_exists='replace', index=False)
df2.to_sql('marketing_data_dictionary', engine, if_exists='replace', index=False)

print("Todos os CSVs foram importados!")

with engine.connect() as conn:
    resultado = conn.execute(text("select table_name from information_schema.tables where table_schema = 'public'"))
    for linha in resultado:
        print(linha[0])
```

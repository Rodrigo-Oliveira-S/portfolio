```python
from sqlalchemy import text
from database import get_engine
from queries import (
    volume_total_saldo, clientes_mais_produtos, 
    faixa_etaria_clientes, tempo_relacionamento, 
    categoria_renda, porcentagem_cancelamento
)
import pandas as pd

engine = get_engine()
```

```python
df1 = pd.read_csv('../raw/bank_churn_data_dictionary.csv')
df2 = pd.read_csv('../raw/bank_churn.csv')

df1.to_sql('bank_churn_data_dictionary', engine, if_exists='replace', index=False)
df2.to_sql('bank_churn', engine, if_exists='replace', index=False)

print('Todas as tabelas foram importadas!')
```

```python
with engine.connect() as conn:
    resultado = conn.execute(text("select table_name from information_schema.tables where table_schema = 'public'"))
    for linha in resultado:
        print(linha[0])
```

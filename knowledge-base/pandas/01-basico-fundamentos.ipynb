### Passo 4 - Resumo Estatístico Numérico

```python
# Gera estatísticas descritivas para todas as colunas numéricas
display(df.describe())
```

O médoto *.describe()* é o principal "raio-x" numérico no início de uma **Análise Exploratória de Dados (EDA).**


### Passo 5 - Análise de contagem e frequência de categorias

O método describe foca apenas em colunas numéricas. Para análisar **colunas de texto/categorias** como categorias ou região, usamos a função *value_counts()* em uma Series

```python
# Contagem de vendas por categoria de produtos
display(df['categoria'].value_counts())
```

```python
# Contagem de vendas por região
display(df['regiao'].value_counts())
```

Permite responder instantaneamente perguntas de negócios sem precisar de agrupamentos complexos, tais como: "*Qual é a categoria mais vendida?*" ou "*Qual região com maior volume de transações?*"

### Passo 6 - Ordenação de Dados

```python
# Ordena o DataFrame pelo valor da venda do maior para o menos, ou vice versa
df_ordenado = df.sort_values(by='valor_venda', ascending=False)

display(df_ordenado)
```

### Passo 7 - Seleção Condicional / Indexação Booleana

```python
# 1. Cria uma máscara booleana (retorna True ou False para cada linha)
mascara = df['valor_venda'] > 1000

# 2. Aplica a máscara no DataFrame para filtrar apenas as linhas True
vendas_altas = df[mascara]

display(vendas_altas)
```

Essa é uma forma nativa do Pandas para filtrar dados. Evita uso de laços *for* do python, garantindo máxima velocidade de processamento, mesmo em bases com milhões de linhas

### Passo 8 - Filtragem Condicional com Múltiplas Condições

```python
# Múltiplas condições: categoria 'Eletronicos' e Região 'Sudeste'
mascara_composta = (df['categoria'] == 'Eletrônicos') & (df['regiao'] == 'Sudeste')

vendas_eletronicas_sudeste = df[mascara_composta]

display(vendas_eletronicas_sudeste)
```

- **Operadores Bitwise(& para E, | para ou):** No python padrão usamos *and* e *or*, mas no pandas **devemos obrigatóriamente usar & e |**. Isso ocorre por que precisamos comparar cada elemento dos arrays veotorizados linha a linha

- **Uso obrigatório do Parênteses:** Em python, o operador & tem prioridade matemática sobre os operadores de comparação (==, >, <). Se não colocar parênteses em cada condição individual *(condicao1) & (condicao2)*, o python gera um erro de sintaxe ou avalia a expressão incorretamente.

### Passo 9 - Indexação Profissional com *loc* e *iloc*

- .loc (Baseado em Rótulos / Nomes): Usa os nomes das colunas e os índices das linhas.

- .iloc (Baseado em Posição Númerica): Usa números de posição inteiros 0, 1, 2, ... tal como python nativo

```python
# 1. Usando .loc: Seleciona linhas de índice 0 a 3, trazendo apenas as colunas 'cliente' e 'valor_venda'
subconjunto_loc = df.loc[0:3, ['cliente', 'valor_venda']]

display(subconjunto_loc)
```

```python
# 2. Usando .iloc: Seleciona as primeiras 3 linhas (índices 0 a 2) e as primeiras 2 colunas (índices 0 e 1)
subconjunto_iloc = df.iloc[0:3, 0:2]

display(subconjunto_iloc)
```

### Passo 10 - Criação e Alteração de Colunas Derivadas

```python
# Cria a coluna 'faturamento' multiplicando valor_venda por quantidade
df['faturamento'] = df['valor_venda'] * df['quantidade']

display(df)
```

### Passo 11 - Exportação de Dados Tratados

```python
# Salva o DataFrame modificado em um novo ficheiro CSV
df.to_csv('~/work-environment/study-files/vendas_limpo.csv', index=False)
```

- **to_csv('caminho/ficheiro.csv'):** Exporta a estrutura bidimensional do DataFrame para um ficheiro de texto separado por vírgulas.

- **index=False (crucial):** Por padrão, o Pandas tenta gravar a coluna de índices numéricos (0, 1, 2...) no ficheiro csv. Definir *index=False* evita que seja criada uma coluna sem nome inútil sempre que reabrir o ficheiro no futuro.
